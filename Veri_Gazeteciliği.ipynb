# **1- Web Scrapping**

```python
# 1. AYARLAR
base_url = "https://www.terorsehitleri.com"
ana_url = "https://www.terorsehitleri.com/sehit" # Liste sayfasının doğru adresi

# Toplanan tüm verileri burada tutacağız
toplanan_tum_linkler = []

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
}

# Başlangıç
sayfa_no = 1
tahmini_sayfa_sayisi = 280 # Sizin belirttiğiniz hedef
guvenlik_siniri = 300      # 280'i geçerse sonsuza gitmesin diye durdurucu

# 2. DÖNGÜ (While True: Liste bitene kadar devam et)
while True:

    # URL Oluşturma (?p=1, ?p=2...)
    guncel_url = f"{ana_url}?p={sayfa_no}"

    # İlerleme durumu
    print(f"📄 [{sayfa_no}/{tahmini_sayfa_sayisi}] Taranıyor...", end=" ")

    try:
        response = requests.get(guncel_url, headers=headers)

        if response.status_code == 200:
            soup = BeautifulSoup(response.content, "html.parser")

            # --- SAYFA İÇİNDEKİ LİNKLERİ BUL ---
            sayfadaki_linkler = []

            # Sitenin yapısına göre tüm linkleri tara
            for link in soup.find_all("a", href=True):
                adres = link['href']

                # Filtre: '/sehit/' ile başlayan ve anlamsız kısa olmayan linkler
                if adres.startswith("/sehit/") and len(adres) > 5:
                    tam_adres = base_url + adres
                    sayfadaki_linkler.append(tam_adres)

            # Sayfa içindeki tekrarları temizle
            sayfadaki_linkler = list(set(sayfadaki_linkler))

            # --- KRİTİK KONTROL NOKTASI ---
            # Eğer sayfadan 0 link çıktıysa, liste bitmiş demektir.
            if len(sayfadaki_linkler) == 0:
                print(f"\n🏁 Sayfa {sayfa_no} boş geldi. Liste sonuna ulaşıldı!")
                break

            # Linkleri ana toplanan_tum_linkler'e ekle
            toplanan_tum_linkler.extend(sayfadaki_linkler)
            print(f"✅ +{len(sayfadaki_linkler)} kişi (Toplam: {len(toplanan_tum_linkler)})")

            # Bir sonraki sayfaya geç
            sayfa_no += 1

            # --- GÜVENLİK KONTROLÜ ---
            if sayfa_no > guvenlik_siniri:
                print(f"\n⚠️ Güvenlik sınırına ({guvenlik_siniri}) gelindi. Döngü zorla durduruluyor.")
                break

            # Siteyi yormamak için kısa bir mola (Çok önemli!)
            time.sleep(1)

        else:
            print(f"❌ Erişim Hatası: {response.status_code}")
            # Hata 404 ise muhtemelen sayfa bitti demektir
            if response.status_code == 404:
                break
            time.sleep(5) # Hata varsa biraz daha uzun bekle

    except Exception as e:
        print(f"\n⚠️ Beklenmedik Hata: {e}")
        time.sleep(3)

# 3. SONUÇ RAPORU
print("\n" + "="*40)
print(f"🎉 TARAMA TAMAMLANDI")
print(f"Taranan Sayfa: {sayfa_no - 1}")
print(f"Toplanan Şehit Profili: {len(toplanan_tum_linkler)}")
# Tekrarları son kez temizle (Farklı sayfalarda aynı kişi varsa)
unique_linkler = list(set(toplanan_tum_linkler))
print(f"Benzersiz Kayıt Sayısı: {len(unique_linkler)}")
print("="*40)

# Örnek Linkler
print("İlk 5 Kayıt:")
for l in unique_linkler[:5]:
    print(l)
```

```python
# 4. AŞAMA: DETAYLI VERİ ÇEKME (TABLO OLUŞTURMA)
print(f"🚀 2. AŞAMA BAŞLIYOR: Detaylı Profil Bilgileri Çekiliyor...")
detayli_veri_listesi = []

# Tüm linkleri gezmek için: unique_linkler
# Test için (sadece ilk 5): unique_linkler[:5]
gezilecek_liste = unique_linkler

for sira, link in enumerate(gezilecek_liste, 1):
    print(f"[{sira}/{len(gezilecek_liste)}] Veri çekiliyor: {link}")

    try:
        r = requests.get(link, headers=headers)
        if r.status_code == 200:
            s_detay = BeautifulSoup(r.content, "html.parser")

            # Her kişi için boş bir kimlik kartı (Sözlük) oluştur
            kisi_karti = {"Profil URL": link}

            # --- İSİM BİLGİSİ ---
            # HTML yapısı karışık olsa bile URL'den ismi temiz bir şekilde alabiliriz
            # Örn: .../sehit/emre-albayrak -> Emre Albayrak
            slug = link.split("/")[-1]
            kisi_adi = slug.replace("-", " ").title()
            kisi_karti["İsim"] = kisi_adi

            # --- TABLO VERİLERİNİ BULMA ---
            # Sizin paylaştığınız HTML yapısındaki satırları (grid-cols-12) buluyoruz
            # Lambda fonksiyonu ile içinde "grid-cols-12" geçen tüm divleri arıyoruz
            bilgi_satirlari = s_detay.find_all("div", class_=lambda x: x and "grid-cols-12" in x and "items-start" in x)

            for satir in bilgi_satirlari:
                # Sol Taraf (Etiket - Label): col-span-4
                etiket_div = satir.find("div", class_=lambda x: x and "col-span-4" in x)
                # Sağ Taraf (Değer - Value): col-span-8
                deger_div = satir.find("div", class_=lambda x: x and "col-span-8" in x)

                if etiket_div and deger_div:
                    etiket = etiket_div.text.strip()
                    deger = deger_div.text.strip()

                    # Veriyi sözlüğe ekle (Örn: kisi_karti["Baba Adı"] = "Ramazan")
                    kisi_karti[etiket] = deger

            # Kişiyi ana listeye ekle
            detayli_veri_listesi.append(kisi_karti)

            # Hızlı istek atıp engellenmemek için
            time.sleep(0.5)

        else:
            print(f"❌ Hata: {r.status_code}")

    except Exception as e:
        print(f"⚠️ Beklenmedik Hata: {e}")

# 5. VERİYİ DATAFRAME'E ÇEVİR VE GÖSTER
print("\n" + "="*40)
if detayli_veri_listesi:
    df = pd.DataFrame(detayli_veri_listesi)

    # Sütunları sizin istediğiniz sıraya diziyoruz
    istenen_sutun_sirasi = [
        "İsim", "Baba Adı", "Anne Adı", "Doğum Tarihi", "Doğum Yeri",
        "Medeni Hali", "Çocuk Sayısı", "Eğitim Durumu", "Şehadet Tarihi",
        "Şehadet Türü", "Görev Yeri", "Şehit Olduğu Yer", "Rütbesi",
        "Defnedildiği Yer", "Şehadet Öyküsü", "Profil URL"
    ]

    # DataFrame'i yeniden indeksleyerek (reindex) eksik sütun varsa NaN yapar, fazlasını atar
    df_final = df.reindex(columns=istenen_sutun_sirasi)

    print(f"✅ İŞLEM TAMAMLANDI! {len(df_final)} satırlık veri seti oluşturuldu.")
    print(df_final.head()) # İlk 5 satırı göster

else:
    print("❌ Hiç veri toplanamadı.")
```

```python
# --- AKILLI DÜZELTME FONKSİYONU ---
def sehir_duzeltici(gelen_veri):
    # 1. Boş veya Çöp Veri Kontrolü
    if pd.isna(gelen_veri) or str(gelen_veri).strip().lower() in [s.lower() for s in silinecekler]:
        return np.nan # Veriyi sil (Boş bırak)

    # 2. Temizlik (Baş harf büyüt, boşluk sil)
    veri = str(gelen_veri).strip()
    veri = veri.title() # Şehir adını baştan Title Case yapıyoruz

    # "Adana, Ceyhan" gibi verilerdeki virgülden sonrasını at
    if "," in veri:
        veri = veri.split(",")[0].strip()
    if "/" in veri:
        veri= veri.split("/")[0].strip()

    # 3. Tam Eşleşme Var mı? (Örn: "Ankara" -> "Ankara")
    # Veri zaten Title Case olduğu için direkt listede arıyoruz
    if veri in standart_sehirler:
        return veri

    # 4. BULANIK EŞLEŞME (Fuzzy Match) -  Ratcliff-Obershelp algoritmasını kullanır. Bu algoritma, iki dizginin ortak dizilerinin uzunluğunu ve sırasını dikkate alır.
    # difflib kütüphanesi en yakın benzerini bulur.
    # cutoff=0.6 demek: %60 ve üzeri benzerlik varsa eşleştir.
    yakin_eslesmeler = difflib.get_close_matches(veri, standart_sehirler, n=1, cutoff=0.7)

    if yakin_eslesmeler:
        # Örn: "Nevşehi̇r" -> "Nevşehir", "İstnbul" -> "İstanbul"
        return yakin_eslesmeler[0]

    return veri # Hiçbir şeye benzemiyorsa olduğu gibi bırak (Daha sonra manuel bakılır)
```

```python
def sutun_temizle_ve_raporla(df, sutun_adi, duzeltici_fonksiyon, referans_liste):
    """
    Belirtilen sütunu temizler, değişiklikleri raporlar ve referans listeye göre kontrol eder.

    Parametreler:
    df (pd.DataFrame): İşlem yapılacak veri çerçevesi.
    sutun_adi (str): Temizlenecek sütunun adı.
    duzeltici_fonksiyon (func): Her bir hücreye uygulanacak temizleme fonksiyonu (sehir_duzeltici).
    referans_liste (list): Geçerli kabul edilen değerlerin listesi (standart_sehirler).
    """

    # 1. Sütun Kontrolü
    if sutun_adi not in df.columns:
        print(f"⚠️ Uyarı: '{sutun_adi}' sütunu veri setinde bulunamadı.")
        return df

    print(f"\n🛠️ '{sutun_adi}' düzeltiliyor, lütfen bekleyin...")

    # 2. Yedekleme (Karşılaştırma için)
    # Performans Notu: .copy() ile bellekte yer ayırıyoruz ama karşılaştırma için gerekli.
    eski_veriler = df[sutun_adi].copy()

    # 3. Fonksiyonu Uygulama
    # apply yerine map kullanmak mümkünse daha hızlıdır ama özel fonksiyonlar için apply kalabilir.
    df[sutun_adi] = df[sutun_adi].apply(duzeltici_fonksiyon)

    # 4. Değişiklik İstatistikleri
    # Hem eski verisi boş olmayan hem de değeri değişenleri say
    degisen_maske = (eski_veriler.notna()) & (eski_veriler != df[sutun_adi])
    degisen_sayisi = degisen_maske.sum()

    silinen_sayisi = df[sutun_adi].isna().sum()

    print(f"   • Düzeltilen Kayıt Sayısı: {degisen_sayisi}")
    print(f"   • Silinen/Boşaltılan Kayıt: {silinen_sayisi}")

    # 5. Son Kontrol (Validasyon)
    print(f"   🔍 '{sutun_adi}' için son kontrol yapılıyor...")

    # Geçici sütun oluşturmadan doğrudan maskeleme ile kontrol (Memory Friendly)
    # Mantık: (Listede YOKSA) VE (Boş DEĞİLSE)
    kontrol_maskesi = (
        (~df[sutun_adi].isin(referans_liste)) &
        (df[sutun_adi].notna())
    )

    hatali_degerler = df.loc[kontrol_maskesi, sutun_adi].unique()

    if len(hatali_degerler) > 0:
        print(f"   ⚠️ Standart dışı kalan {len(hatali_degerler)} farklı değer tespit edildi:")
        print("   " + "-" * 40)
        for deger in hatali_degerler:
            # Bu değerden kaç tane olduğunu hesapla
            adet = len(df[df[sutun_adi] == deger])
            print(f"   • '{deger}' ({adet} kayıt)")
    else:
        print(f"   ✅ Harika! '{sutun_adi}' sütunundaki tüm veriler standart listeye uygun.")

    return df
```

```python
def prepare_city_counts(df, column_name):
    """
    Belirtilen sütundaki şehir sayımlarını hazırlar ve sıralar.

    Parametreler:
    df (pd.DataFrame): İşlem yapılacak veri çerçevesi.
    column_name (str): Şehir verilerini içeren sütun adı.

    Döndürür:
    pd.DataFrame: 'Sehir' ve 'Kisi_Sayisi' sütunlarına sahip, şehre göre sıralanmış DataFrame.
    """
    # 1. Sayım yap
    sehir_sayilari = df[column_name].value_counts()

    # 2. Tabloya çevir
    sehir_tablosu = sehir_sayilari.reset_index()

    # 3. Sütun isimlerini düzelt
    sehir_tablosu.columns = ['Sehir', 'Kisi_Sayisi']

    # 4. Şehir ismine göre alfabetik sırala
    sehir_tablosu = sehir_tablosu.sort_values(by='Sehir')

    return sehir_tablosu

# 1. Doğum Yeri için fonksiyonu kullan
sehir_tablosu = prepare_city_counts(df_final, 'Doğum Yeri')

# Sonucu göster
print(sehir_tablosu.head(10))
print("-" * 30)
print(f"Toplam benzersiz şehir sayısı: {len(sehir_tablosu)}")
```

```python
# Boş verileri atıyoruz
tarih = df_final_copy['Şehadet Tarihi'].dropna().astype(str)

# Tarih formatlama (Gün kısmını atma - .str[3:] yöntemi)
ay_yil_serisi = tarih.str[3:]

# DÜZELTME: "01.0001" gibi hatalı yılları "01.2001" ile değiştir
ay_yil_serisi = ay_yil_serisi.str.replace('01.0001', '01.2001', regex=False)

# Sayma ve DataFrame Oluşturma
df_sonuc = ay_yil_serisi.value_counts().reset_index()
df_sonuc.columns = ['Tarih', 'Sayi']

# Tarih Referansı oluşturma (Sıralama ve Grafik için)
# errors='coerce' ile hatalı formatları NaT'a çeviriyoruz
df_sonuc['Tarih_Referans'] = pd.to_datetime(df_sonuc['Tarih'], format='%m.%Y', errors='coerce')

# Hatalı tarihleri (NaT) temizle
df_sonuc = df_sonuc.dropna(subset=['Tarih_Referans'])

# FİLTRELEME: 1993 yılından önceki kayıtları sil
df_sonuc = df_sonuc[df_sonuc['Tarih_Referans'].dt.year >= 1993]

# SIRALAMA (DÜZELTME YAPILDI)
# Tarih referansına göre sıralıyoruz ancak sütunu ARTIK SİLMİYORUZ.
# Çünkü aşağıda grafikte x ekseni olarak kullanılacak.
df_sonuc = df_sonuc.sort_values('Tarih_Referans').reset_index(drop=True)

# Sonucu konsola basalım (Kontrol amaçlı)
print(df_sonuc.head())
```

```python
# ---------------------------------------------------------
# 2. PLOTLY İLE GÖRSELLEŞTİRME
# ---------------------------------------------------------

# Plotly'nin zaman eksenini doğru kullanabilmesi için 'Tarih_Referans' sütununu
# x ekseni olarak kullanacağız (Datetime formatında olduğu için).

fig_sehit_aylik_ts = px.line(
    df_sonuc,
    x='Tarih_Referans',
    y='Sayi',
    title='Yıllara Göre Şehit Sayıları (1993 ve Sonrası - Aylık Dağılım)',
    labels={'Tarih_Referans': 'Tarih', 'Sayi': 'Şehit Sayısı'},
    markers=True,  # Veri noktalarını nokta olarak göster
    template="plotly_white" # Temiz beyaz arka plan
)

# Çizgi Rengi ve Stili
fig_sehit_aylik_ts.update_traces(line_color='#d62728', marker=dict(size=6)) # Kırmızı renk

# Ekseni ve Slider Ayarları
fig_sehit_aylik_ts.update_layout(
    xaxis=dict(
        rangeselector=dict(
            buttons=list([
                dict(count=1, label="1 Yıl", step="year", stepmode="backward"),
                dict(count=5, label="5 Yıl", step="year", stepmode="backward"),
                dict(count=10, label="10 Yıl", step="year", stepmode="backward"),
                dict(step="all", label="Tümü")
            ])
        ),
        rangeslider=dict(visible=True), # Alt tarafa kaydırma çubuğu ekler
        type="date"
    ),
    yaxis=dict(title='Şehit Sayısı'),
    hovermode="x unified" # Fare ile üzerine gelince o aydaki tüm bilgiyi göster
)

# Grafiği Göster
fig_sehit_aylik_ts.show()
```

```python
# ---------------------------------------------------------
# 2. YILLARA GÖRE GRUPLAMA (YENİ KISIM)
# ---------------------------------------------------------

# Tarih_Referans sütunundan "Yıl" bilgisini çekip grupluyoruz ve sayıları topluyoruz.
df_yillik = df_sonuc.groupby(df_sonuc['Tarih_Referans'].dt.year)['Sayi'].sum().reset_index()

# Sütun isimlerini düzenleyelim
df_yillik.columns = ['Yıl', 'Toplam_Şehit_Sayısı']

print("✅ Yıllık Özet Tablo:")
print(df_yillik.head())
```

```python
# ---------------------------------------------------------
# 3. YILLIK GRAFİK (PLOTLY)
# ---------------------------------------------------------

# Yıllık verilerde genellikle "Bar Chart" (Sütun Grafik) daha net görünür.
fig_sehit_yillik_bar = px.bar(
    df_yillik,
    x='Yıl',
    y='Toplam_Şehit_Sayısı',
    title='Yıllara Göre Toplam Şehit Sayıları (1993-2025)',
    labels={'Yıl': 'Yıl', 'Toplam_Şehit_Sayısı': 'Şehit Sayısı'},
    text='Toplam_Şehit_Sayısı', # Sütunların üzerine sayıları yaz
    template="plotly_white"
)

# Görsel Ayarlar
fig_sehit_yillik_bar.update_traces(
    marker_color='#d62728', # Kırmızı
    textposition='outside'  # Sayıları sütunun tepesine koy
)

fig_sehit_yillik_bar.update_layout(
    xaxis=dict(
        tickmode='linear', # Tüm yılları eksende tek tek göster
        dtick=1            # 1'er yıl arayla
    ),
    yaxis=dict(title='Şehit Sayısı')
)

fig_sehit_yillik_bar.show()
```
